```python
import pandas as pd
from sklearn.model_selection import train_test_split

loaded_df  = pd.read_csv('./tornado_data_for_model_training.csv')

df = loaded_df.dropna(subset=["ef_category"]).reset_index(drop=True)
print(len(df))
```

```python
from sklearn.model_selection import train_test_split

RNG = 12345

df["ef_class"] = df["ef_category"].astype(int)
df_dropped = df.drop(["ef_category", 'Unnamed: 0'], axis=1)

df["ef_binary"] = (df["ef_class"] >= 2).astype(int)

df_trainval, df_test = train_test_split(
    df_dropped,
    test_size=0.15,
    stratify=df["ef_binary"],
    random_state=RNG,
)

df_train, df_val = train_test_split(
    df_trainval,
    test_size=0.17647058823529413,  # 0.15/0.85 to make final 70/15/15
    stratify=df_trainval["ef_binary"],
    random_state=RNG,
)

# 4) Save split CSVs
df_train.to_csv("train.csv", index=False)
df_val.to_csv("val.csv", index=False)
df_test.to_csv("test.csv", index=False)
```
