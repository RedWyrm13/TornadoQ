```python
import pandas as pd

df_1 = pd.read_csv("./train.csv")
df_2 = pd.read_csv("./val.csv")
df_3 = pd.read_csv("./test.csv")

df = pd.concat([df_1, df_2, df_3])


```

```python
from sklearn.model_selection import train_test_split

def split_three_equal(df, seed=42):
    # First split: 1/3 test, 2/3 remaining
    df_temp, df_test = train_test_split(df, test_size=1/3, random_state=seed, shuffle=True)

    # Second split: split remaining into train and val equally
    df_train, df_val = train_test_split(df_temp, test_size=1/2, random_state=seed, shuffle=True)

    return df_train, df_val, df_test

df_train, df_val, df_test = split_three_equal(df)
```

```python
df_train.to_csv("third_train.csv")
df_val.to_csv("third_val.csv")
df_test.to_csv("third_test.csv")
```
